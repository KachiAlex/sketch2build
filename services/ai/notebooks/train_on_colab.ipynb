## 3. Train Vision Encoder (Sketch → Room Graph)

Trains the cross-attention room detection head on top of CLIP ViT-Large.
- **Epochs:** 100 (with early stopping, patience=20)
- **Batch size:** 4 (T4 VRAM optimized)
- **Learning rate:** 1e-4 with cosine warmup
- **Estimated time:** ~2.5 hours on T4

```python
# Optional: Login to W&B for training visualization
# (Skip if you don't have a W&B account — training will still work)
USE_WANDB = False  # Set to True if you want W&B logging

if USE_WANDB:
    !pip install -q wandb
    import wandb
    wandb.login()  # You'll need to paste your API key
else:
    print('Skipping W&B. Set USE_WANDB=True and run wandb.login() to enable.')
```

```python
import os, sys, time, json
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, random_split
from torch.optim import AdamW
from torch.optim.lr_scheduler import CosineAnnealingLR, LinearLR
from transformers import CLIPProcessor
import structlog

sys.path.insert(0, '/content/sketch2build/services/ai')

from src.models.vision.sketch_encoder import SketchEncoderModel
from src.training.data.dataset import SketchDataset

structlog.get_logger().info('Starting vision encoder training')

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
BATCH_SIZE = 4  # T4 has 16GB VRAM, CLIP ViT-Large needs small batches
EPOCHS = 100
LR = 1e-4
WARMUP_STEPS = 500
PATIENCE = 20
MAX_GRAD_NORM = 1.0
VAL_SPLIT = 0.1

OUTPUT_DIR = f'{DRIVE_DIR}/vision_encoder'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Create model
print('Loading CLIP ViT-Large + cross-attention room detection head...')
model = SketchEncoderModel(
    pretrained_model='openai/clip-vit-large-patch14',
    freeze_backbone=False,
).to(DEVICE)
print(f'Model loaded on {DEVICE}')
print(f'Total parameters: {sum(p.numel() for p in model.parameters()):,}')
print(f'Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}')

# Create processor
processor = CLIPProcessor.from_pretrained('openai/clip-vit-large-patch14')

# Create dataset
print('Loading dataset...')
full_dataset = SketchDataset(
    metadata_path='/content/data/synthetic/metadata.json',
    image_dir='/content/data/synthetic',
    processor=processor,
    split='train',
    max_rooms=20,
)
print(f'Dataset size: {len(full_dataset)}')

# Split train/val
val_size = int(len(full_dataset) * VAL_SPLIT)
train_size = len(full_dataset) - val_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])
print(f'Train: {train_size}, Val: {val_size}')
```

```python
# Collate function
def collate_fn(batch):
    max_rooms = 20
    pixel_values = torch.stack([item['pixel_values'] for item in batch])
    B = len(batch)
    room_presence = torch.zeros(B, max_rooms)
    room_types = torch.full((B, max_rooms), -1, dtype=torch.long)
    bboxes = torch.zeros(B, max_rooms, 4)
    adjacency = torch.zeros(B, max_rooms, max_rooms)
    mask = torch.zeros(B, max_rooms, dtype=torch.bool)
    
    for b, item in enumerate(batch):
        mask[b] = item['mask']
        room_types[b] = item['room_types']
        bboxes[b] = item['bboxes']
        room_presence[b] = item['mask'].float()
        adjacency[b] = item['adjacency']
    
    return {
        'pixel_values': pixel_values,
        'room_presence': room_presence,
        'room_types': room_types,
        'bboxes': bboxes,
        'adjacency': adjacency,
        'mask': mask,
    }

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, 
                          num_workers=2, collate_fn=collate_fn, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=2, collate_fn=collate_fn, pin_memory=True)

# Optimizer
optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
total_steps = len(train_loader) * EPOCHS
warmup_scheduler = LinearLR(optimizer, start_factor=0.01, total_iters=WARMUP_STEPS)
cosine_scheduler = CosineAnnealingLR(optimizer, T_max=total_steps - WARMUP_STEPS)

print(f'Total steps: {total_steps}')
print(f'Steps per epoch: {len(train_loader)}')
```

```python
# Training loop
best_val_loss = float('inf')
patience_counter = 0
global_step = 0

# Check for existing checkpoint to resume
ckpt_path = os.path.join(OUTPUT_DIR, 'best.pt')
if os.path.exists(ckpt_path):
    print(f'Resuming from checkpoint: {ckpt_path}')
    epoch_start, best_val_loss = model.load_checkpoint(ckpt_path, optimizer=optimizer, device=DEVICE)
    print(f'Resumed at epoch {epoch_start}, best val loss: {best_val_loss:.4f}')
else:
    epoch_start = 0
    print('Starting fresh training.')

print(f'\nStarting training for {EPOCHS} epochs...')
print(f'Estimated time: ~{len(train_loader) * EPOCHS * 0.15 / 3600:.1f} hours on T4')

train_losses = []
val_losses = []

for epoch in range(epoch_start, EPOCHS):
    model.train()
    epoch_loss = 0.0
    epoch_presence = 0.0
    epoch_type = 0.0
    epoch_bbox = 0.0
    epoch_adj = 0.0
    num_batches = 0
    start_time = time.time()
    
    for batch_idx, batch in enumerate(train_loader):
        pixel_values = batch['pixel_values'].to(DEVICE)
        targets = {k: v.to(DEVICE) for k, v in batch.items() if k != 'pixel_values'}
        
        optimizer.zero_grad()
        outputs = model(pixel_values)
        losses = model.compute_loss(outputs, targets)
        losses['total'].backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
        optimizer.step()
        
        epoch_loss += losses['total'].item()
        epoch_presence += losses['presence'].item()
        epoch_type += losses['type'].item()
        epoch_bbox += losses['bbox'].item()
        epoch_adj += losses['adjacency'].item()
        num_batches += 1
        global_step += 1
        
        # LR scheduler
        if global_step <= WARMUP_STEPS:
            warmup_scheduler.step()
        else:
            cosine_scheduler.step()
        
        if (batch_idx + 1) % 50 == 0:
            elapsed = time.time() - start_time
            print(f'  Epoch {epoch+1}/{EPOCHS} | Batch {batch_idx+1}/{len(train_loader)} | '
                  f'Loss: {losses["total"].item():.4f} | {elapsed:.0f}s')
    
    # Validation
    model.eval()
    val_loss = 0.0
    val_batches = 0
    correct_types = 0
    total_rooms = 0
    
    with torch.no_grad():
        for batch in val_loader:
            pixel_values = batch['pixel_values'].to(DEVICE)
            targets = {k: v.to(DEVICE) for k, v in batch.items() if k != 'pixel_values'}
            outputs = model(pixel_values)
            losses = model.compute_loss(outputs, targets)
            val_loss += losses['total'].item()
            val_batches += 1
            
            valid = targets['room_types'] >= 0
            if valid.any():
                pred_types = outputs['room_type_logits'][valid].argmax(dim=-1)
                true_types = targets['room_types'][valid]
                correct_types += (pred_types == true_types).sum().item()
                total_rooms += valid.sum().item()
    
    avg_train_loss = epoch_loss / max(num_batches, 1)
    avg_val_loss = val_loss / max(val_batches, 1)
    type_acc = correct_types / max(total_rooms, 1)
    epoch_time = time.time() - start_time
    
    train_losses.append(avg_train_loss)
    val_losses.append(avg_val_loss)
    
    print(f'\nEpoch {epoch+1}/{EPOCHS} DONE | Train Loss: {avg_train_loss:.4f} | '
          f'Val Loss: {avg_val_loss:.4f} | Type Acc: {type_acc*100:.1f}% | {epoch_time:.0f}s')
    
    # Save best checkpoint
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        patience_counter = 0
        model.save_checkpoint(
            os.path.join(OUTPUT_DIR, 'best.pt'),
            optimizer=optimizer,
            epoch=epoch + 1,
            best_loss=best_val_loss,
        )
        print(f'  ★ New best model saved! Val loss: {best_val_loss:.4f}')
    else:
        patience_counter += 1
        print(f'  No improvement. Patience: {patience_counter}/{PATIENCE}')
    
    # Periodic checkpoint
    if (epoch + 1) % 5 == 0:
        model.save_checkpoint(
            os.path.join(OUTPUT_DIR, f'checkpoint_epoch_{epoch+1}.pt'),
            optimizer=optimizer,
            epoch=epoch + 1,
            best_loss=best_val_loss,
        )
    
    # Early stopping
    if patience_counter >= PATIENCE:
        print(f'\nEarly stopping at epoch {epoch+1}! Best val loss: {best_val_loss:.4f}')
        break

print(f'\n✅ Vision encoder training complete! Best val loss: {best_val_loss:.4f}')
print(f'Checkpoints saved to: {OUTPUT_DIR}')
```

## 4. Train Layout Diffusion Model (Room Graph → Floor Plan)

Trains the full U-Net diffusion model conditioned on room graphs.
- **Epochs:** 100 (with early stopping, patience=30)
- **Batch size:** 2 (T4 VRAM optimized — U-Net is memory-heavy)
- **Learning rate:** 1e-4 with cosine warmup
- **Estimated time:** ~7-8 hours on T4

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, random_split
from torch.optim import AdamW
from torch.optim.lr_scheduler import CosineAnnealingLR, LinearLR
import numpy as np
import time, os

from src.models.layout.diffusion import LayoutDiffusionModel
from src.training.data.dataset import LayoutDataset

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
BATCH_SIZE = 2  # U-Net is memory-heavy, keep batch small on T4
EPOCHS = 100
LR = 1e-4
WARMUP_STEPS = 500
PATIENCE = 30
MAX_GRAD_NORM = 1.0
VAL_SPLIT = 0.1

OUTPUT_DIR = f'{DRIVE_DIR}/layout_diffusion'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Create model
print('Creating Layout Diffusion Model (U-Net + DDPM)...')
model = LayoutDiffusionModel(num_train_timesteps=1000).to(DEVICE)
print(f'Model on {DEVICE}')
print(f'Total parameters: {sum(p.numel() for p in model.parameters()):,}')

# Create dataset
print('Loading dataset...')
full_dataset = LayoutDataset(
    metadata_path='/content/data/synthetic/metadata.json',
    image_dir='/content/data/synthetic',
    split='train',
    image_size=256,
)
print(f'Dataset size: {len(full_dataset)}')

val_size = int(len(full_dataset) * VAL_SPLIT)
train_size = len(full_dataset) - val_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])
print(f'Train: {train_size}, Val: {val_size}')
```

```python
# Collate function for layout diffusion
def layout_collate_fn(batch):
    max_rooms = 20
    images = torch.stack([item['image'] for item in batch])
    B = len(batch)
    room_types = torch.full((B, max_rooms), 0, dtype=torch.long)
    bboxes = torch.zeros(B, max_rooms, 4)
    adjacency = torch.zeros(B, max_rooms, max_rooms)
    mask = torch.zeros(B, max_rooms, dtype=torch.bool)
    
    for b, item in enumerate(batch):
        rooms = item.get('rooms', [])
        for i, room in enumerate(rooms[:max_rooms]):
            room_types[b, i] = room.get('type_idx', 0)
            bbox = room.get('bbox', [0, 0, 0, 0])
            if len(bbox) == 4:
                bboxes[b, i] = torch.tensor(bbox, dtype=torch.float32)
            mask[b, i] = True
        
        adj = item.get('adjacency', [])
        for pair in adj:
            if isinstance(pair, (list, tuple)) and len(pair) >= 2:
                a, c = int(pair[0]), int(pair[1])
                if a < max_rooms and c < max_rooms:
                    adjacency[b, a, c] = 1.0
                    adjacency[b, c, a] = 1.0
    
    return {
        'images': images,
        'room_types': room_types,
        'bboxes': bboxes,
        'adjacency': adjacency,
        'mask': mask,
    }

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=2, collate_fn=layout_collate_fn, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=2, collate_fn=layout_collate_fn, pin_memory=True)

optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
total_steps = len(train_loader) * EPOCHS
warmup_scheduler = LinearLR(optimizer, start_factor=0.01, total_iters=WARMUP_STEPS)
cosine_scheduler = CosineAnnealingLR(optimizer, T_max=total_steps - WARMUP_STEPS)

print(f'Total steps: {total_steps}')
print(f'Steps per epoch: {len(train_loader)}')
print(f'Estimated time: ~{len(train_loader) * EPOCHS * 0.5 / 3600:.1f} hours on T4')
```

```python
# Training loop
best_val_loss = float('inf')
patience_counter = 0
global_step = 0

# Check for existing checkpoint
ckpt_path = os.path.join(OUTPUT_DIR, 'best.pt')
if os.path.exists(ckpt_path):
    print(f'Resuming from checkpoint: {ckpt_path}')
    epoch_start, best_val_loss = model.load_checkpoint(ckpt_path, optimizer=optimizer, device=DEVICE)
    print(f'Resumed at epoch {epoch_start}, best val loss: {best_val_loss:.4f}')
else:
    epoch_start = 0
    print('Starting fresh training.')

print(f'\nStarting layout diffusion training for {EPOCHS} epochs...')

layout_train_losses = []
layout_val_losses = []

for epoch in range(epoch_start, EPOCHS):
    model.train()
    epoch_loss = 0.0
    num_batches = 0
    start_time = time.time()
    
    for batch_idx, batch in enumerate(train_loader):
        images = batch['images'].to(DEVICE)
        room_types = batch['room_types'].to(DEVICE)
        bboxes = batch['bboxes'].to(DEVICE)
        adjacency = batch['adjacency'].to(DEVICE)
        mask = batch['mask'].to(DEVICE)
        
        optimizer.zero_grad()
        loss = model(images, room_types, bboxes, adjacency, mask)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
        optimizer.step()
        
        epoch_loss += loss.item()
        num_batches += 1
        global_step += 1
        
        if global_step <= WARMUP_STEPS:
            warmup_scheduler.step()
        else:
            cosine_scheduler.step()
        
        if (batch_idx + 1) % 100 == 0:
            elapsed = time.time() - start_time
            print(f'  Epoch {epoch+1}/{EPOCHS} | Batch {batch_idx+1}/{len(train_loader)} | '
                  f'Loss: {loss.item():.4f} | {elapsed:.0f}s')
    
    # Validation
    model.eval()
    val_loss = 0.0
    val_batches = 0
    
    with torch.no_grad():
        for batch in val_loader:
            images = batch['images'].to(DEVICE)
            room_types = batch['room_types'].to(DEVICE)
            bboxes = batch['bboxes'].to(DEVICE)
            adjacency = batch['adjacency'].to(DEVICE)
            mask = batch['mask'].to(DEVICE)
            loss = model(images, room_types, bboxes, adjacency, mask)
            val_loss += loss.item()
            val_batches += 1
    
    avg_train_loss = epoch_loss / max(num_batches, 1)
    avg_val_loss = val_loss / max(val_batches, 1)
    epoch_time = time.time() - start_time
    
    layout_train_losses.append(avg_train_loss)
    layout_val_losses.append(avg_val_loss)
    
    print(f'\nEpoch {epoch+1}/{EPOCHS} DONE | Train Loss: {avg_train_loss:.4f} | '
          f'Val Loss: {avg_val_loss:.4f} | {epoch_time:.0f}s')
    
    # Save best
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        patience_counter = 0
        model.save_checkpoint(
            os.path.join(OUTPUT_DIR, 'best.pt'),
            optimizer=optimizer,
            epoch=epoch + 1,
            best_loss=best_val_loss,
        )
        print(f'  ★ New best model saved! Val loss: {best_val_loss:.4f}')
    else:
        patience_counter += 1
        print(f'  No improvement. Patience: {patience_counter}/{PATIENCE}')
    
    # Periodic checkpoint + sample generation
    if (epoch + 1) % 10 == 0:
        model.save_checkpoint(
            os.path.join(OUTPUT_DIR, f'checkpoint_epoch_{epoch+1}.pt'),
            optimizer=optimizer,
            epoch=epoch + 1,
            best_loss=best_val_loss,
        )
        # Generate a sample
        model.eval()
        with torch.no_grad():
            sample = model.generate(
                room_types=room_types[:1],
                bboxes=bboxes[:1],
                adjacency=adjacency[:1],
                mask=mask[:1],
                shape=(1, 3, 256, 256),
                num_inference_steps=20,
            )
            sample_img = sample[0].cpu()
            sample_img = ((sample_img + 1) / 2 * 255).clamp(0, 255).byte()
            from PIL import Image
            Image.fromarray(sample_img.permute(1, 2, 0).numpy()).save(
                os.path.join(OUTPUT_DIR, f'sample_epoch_{epoch+1}.png')
            )
            print(f'  Sample saved: sample_epoch_{epoch+1}.png')
    
    if patience_counter >= PATIENCE:
        print(f'\nEarly stopping at epoch {epoch+1}! Best val loss: {best_val_loss:.4f}')
        break

print(f'\n✅ Layout diffusion training complete! Best val loss: {best_val_loss:.4f}')
print(f'Checkpoints saved to: {OUTPUT_DIR}')
```

## 5. Generate Sample Designs

Test the trained models by generating floor plans from room graphs.
